````python
"""
1.  Get user input using **`input(“Enter your age: ”)`**. If user is 18 or older, give feedback: **`You are old enough to drive`**. If below 18 give feedback to **`You need #number more years to learn to drive`**.

Output:
```py
    Enter your age: 30
    You are old enough to learn to drive.
    Output:
    Enter your age: 15
    You need 3 more years to learn to drive.
```
"""

age = int(input("Enter your age: "))

if age>= 18:
    print("You are old enough to drive")
else:
    print(f"You need {18 - age} more years to learn to drive")
````

````python
"""
2.  Compare the values of my_age and your_age using **`if-else`**. Who is older (me or you)? 
Use **`input(“Enter your age: ”)`** to get the age as input. You can use a nested condition to print **`'year'`** 
for 1 year difference in age, **`'years'`** for bigger differences, and a custom text if **`my_age = your_age`**.


Output:

```py
    Enter your age: 30
    You are 5 years older than me.
```
"""

my_age = 20
your_age = int(input("Enter Your age: "))
dif = abs(my_age - your_age)
sentence = f"{dif} years" if dif>1 else f"a year"

if my_age > your_age:
    print(f"I'm older than you by {sentence}")
else:
    print(f"I'm younger than you by {sentence}")
````

````python
"""
3. Check if the season is Autumn, Winter, Spring or Summer.
    - If the user input is:
        - September, October or November, the season is Autumn.
        - December, January or February, the season is Winter.
        - March, April or May, the season is Spring
        - June, July or August, the season is Summer

"""
months_dect = {
    1: 'january', 2: 'february', 3: 'march', 4: 'april',
    5: 'may', 6: 'june', 7: 'july', 8: 'august',
    9: 'september', 10: 'october', 11: 'november', 12: 'december'
}

month = int(input("Enter a month number in range (1-12): "))

while month not in range(1, 13):
    month = int(input("Invalid. Enter the month num in range (1-12): "))

if month in range (9, 12): 
    print(f"The month {months_dect[month]} is in Autumn")
elif month in {12, 1, 2}: 
    print(f"The month {months_dect[month]} is in Winter")
elif month in range (3, 6): 
    print(f"The month {months_dect[month]} is in Spring")
else: 
    print(f"The month {months_dect[month]} is in Summer")
````

# Part 2: Dictionary & Conditions Practice

You are given a dictionary called `person` that contains information about an individual.  
Your task is to perform the following operations step-by-step:

##  The Dictionary:

```python
person = {
    'first_name': 'Milaan',
    'last_name': 'Parmar',
    'age': 96,
    'country': 'Finland',
    'is_marred': True,
    'skills': ['Python', 'Matlab', 'R', 'C', 'C++'],
    'address': {
        'street': 'Space street',
        'zipcode': '02210'
    }
}
```

````python
"""
### 1. Check for `skills` key
- If the `skills` key exists in the dictionary:
  - Print the **middle skill** from the list of skills.
"""

person = {
    'first_name': 'Milaan',
    'last_name': 'Parmar',
    'age': 96,
    'country': 'Finland',
    'is_marred': True,
    'skills': ['Python', 'Matlab', 'R', 'C', 'C++'],
    'address': {
        'street': 'Space street',
        'zipcode': '02210'
    }
}
print(person['skills'][len(person['skills'])//2]) if person['skills'] else print(" There is no skills")
````

````python
"""
### 2. Check for `'Python'` in skills
- Still inside the same condition:
  - Check if `'Python'` is in the person's skills list.
  - Print `True` if found, otherwise `False`.
"""
exist = False
for skill in person['skills']:
    if skill.lower() == "python":
        print("python exists")
        exist += 1
        break
if exist == False:
    print("python does not exist")
````

````python
"""
### 3. Based on the combination of skills:
  - If the person has **only** `['Python', 'Matlab']` → print: `He knows machine learning`
  - If the person has both `'Python'` and `'R'` → print: `He knows statistics`
  - If the person has both `'C'` and `'C++'` → print: `He knows software development`
  - Else → print: `unknown title`
"""
skills = []

if ("Python" in person['skills']) and 'Matlab' in person['skills']:
    skills.append("He knows machine learning")
if ("Python" in person['skills']) and ("R" in person['skills']):
    skills.append("He knows statistics")
if ("C" in person['skills']) and ("C++" in person['skills']):
    skills.append("He knows software development")

if skills:    
    for i in skills:
        print(i)
else: 
    print("unknown title")
````

````python
""" 
### 4. Print location and marital status
- If the person is married **and** lives in **Finland**, print:
```py
    Milaan Parmar lives in Finland. He is married.
```
"""

print(f"{person['first_name']} lives in {person['country']}")
print(f"{person['first_name']} is is married" if person['is_marred'] else "not married" )

if person['country'] == 'Finland' and person['is_marred']:
    print(f"{person['first_name']} lives in Finland, and he is married")
````
