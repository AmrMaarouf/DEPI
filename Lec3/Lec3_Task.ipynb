````python
#4. Write a function called **`check_season`**, it takes a month parameter and 
# returns the season: Autumn, Winter, Spring or Summer.

def season_finder(month):
    if month > 12 or month < 1:
        return "please enter a num between 1 and 12"
    elif month in (3,4,5):
        return "Spring"        
    elif month in (6,7,8):
        return "Summer"        
    elif month in (9,10,11):
        return "Autumn"
    else:
        return "Winter"

print(season_finder(-1))
print(season_finder(13))
print(season_finder(5))
print(season_finder(1))
````

````python
# 5. Write a function called **`calculate_slope`** which return the slope of a linear equation

def calculate_slope(x1, x2, y1, y2):
    if x1 == x2:
        return "undefined"
    else:
        return (y2 - y1) / (x2 - x1)

print(calculate_slope(1,2,3,4))
print(calculate_slope(1,1,3,4))
````

````python
#6. Quadratic equation is calculated as follows: **ax² + bx + c = 0**. 
# Write a function which calculates solution set of a quadratic equation, **`solve_quadratic_eqn`**.

import math 

def quad_equation_sol(a, b, c):
    
    
    if a == 0:
        return "not a quadratic equation"
    
    discriminant = (b**2-4*a*c)
    
    if discriminant < 0:
        return "No real solution"
    
    elif discriminant == 0:
        return -b/(2*a)
    
    return [(-b + math.sqrt(discriminant))/(2*a),
            (-b - math.sqrt(discriminant))/(2*a)]

print(quad_equation_sol(4,3,2))
print(quad_equation_sol(1,-3,2))
print(quad_equation_sol(1,-4,4))
````

````python
# 1- Declare a function named **`capitalize_list_items`**. 
# It takes a list as a parameter and it returns a capitalized list of items

def capitalize_list_items(words):
    capitalized_list = []
    for word in words:
        capitalized_list.append(word.capitalize())
    return capitalized_list

a_list = ["name", "amr", "Hello"]
print(capitalize_list_items(a_list))
````

````python
# 3. Declare a function named **`reverse_list`**. 
# It takes an array as a parameter and it returns the reverse of the array (use loops).

def reverse_list(items):
    reversed_items = [] 
    for i in items:
        reversed_items.insert(0, i)
    return reversed_items
    
print(reverse_list([1, 2, 3, 4, 5]))
print(reverse_list(["A", "B", "C"]))
````

````python
# 4.  Declare a function named **`evens_and_odds`**. 
# It takes a positive integer as parameter and it counts number of evens and odds in the number.

def evens_and_odds(num):
    even = 0
    odd = 0
    for i in range(num+1):
        if i%2:
            odd += 1
        else:
            even += 1
    return f"There ara {even} even numbers, and {odd} odd numbers"


print(evens_and_odds(100))
````

````python
# 5. Write different functions which take lists. 
# They should **`calculate_mean`**, **`calculate_median`**, **`calculate_mode`**

def calculate_mean(nums):
    sum = 0
    if len(nums) == 0:
        return "there is no items in this list"
    for i in nums:
        sum += i
    return sum/len(nums)

def calculate_median(nums):
    if len(nums) == 0:
        return "there is no items in this list"
    sorted_list = sorted(nums)
    
    if len(sorted_list)%2:
        return sorted_list[len(sorted_list)//2]
    else:
         return (sorted_list[len(sorted_list)//2] +  sorted_list[len(sorted_list)//2-1])/2
     
def calculate_mode(nums):
    if len(nums) == 0:
            return "there is no items in this list"
    nums_dict = {} 
    for i in nums:
        found = False
        for num in nums_dict:
            if i == num:
                nums_dict[num] +=1
                found = True
                break
        if not found:
            nums_dict[i] = 1
    highest_num = None
    highest_count = 0
    
    for val, count in nums_dict.items():
        if count > highest_count:
            highest_count = count
            highest_num = val

    return highest_num


nums = [6, 3, 7, 1, 2, 9, 2, 6, 10, 1]
print(calculate_mean(nums))
print(calculate_median(nums))
print(calculate_mode(nums))
````
