```python
class Person:
    def __init__(self, name, age):
            self.name = name
            self.age = age
            
    def display_name(self):
        print(f"This Person is called {self.name}")
    def display_age(self):
        print(f"{self.name}'s age is {self.age}")

class Student(Person):
    
    def __init__(self, name, age, a_class, GPA):
        super().__init__(name, age)
        self.GPA = GPA
        self.a_class = a_class
        
    def display_GPA(self):
            print(f"{self.name}'s GPA is {self.GPA}")
    def display_class(self):
            print(f"{self.name}'s class is {self.a_class}")

obj1 = Student("Amr", 20, "ISM5", 3.77)
obj2 = Student("ِAhmed", 25, "ISM5", 3.96)

print(obj1)
obj1.display_name()
obj1.display_age()
obj1.display_GPA()
obj1.display_class()
print(obj2)
obj2.display_name()
obj2.display_age()
obj2.display_GPA()
obj2.display_class()

```
