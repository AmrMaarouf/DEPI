```python
car_df = pd.read_csv("car-sales.csv")
car_df
```

```python
car_df.info()
```

```python
car_df["Price"] = (
    car_df["Price"]
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
    .astype(float)
)
```

```python
car_df.info()
```

```python
car_df
```

```python
make_count = car_df["Make"].value_counts()
make_count
```

```python
make_count = car_df["Make"].value_counts()
make_count 
plt.bar(make_count.index, make_count.values)
plt.xlabel("Make")
plt.ylabel("Count")
plt.title("Count of Cars by Make")
```

```python
car_df
```

```python
Colour_count = car_df["Colour"].value_counts()
Colour_count 
plt.bar(Colour_count.index, Colour_count.values)
plt.xlabel("Colour")
plt.ylabel("Count")
plt.title("Count of Cars by Colour")
```
