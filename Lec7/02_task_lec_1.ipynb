# Exploratory Data Analysis Task

## Dataset

**Customer Shopping Data**

Your task is to explore the dataset and answer the following questions using **Pandas** and **Matplotlib**.

---

# Part 1: Data Understanding

1. How many rows and columns are in the dataset?
2. What are the column names?
3. What is the data type of each column?
4. Are there any missing values?
5. Are there any duplicated rows?
6. What is the minimum, maximum, and average customer age?
7. What is the minimum, maximum, and average product price?
8. What are the unique product categories?
9. What payment methods are available?
10. What shopping malls are included in the dataset?

---

# Part 2: Comparison Analysis

Use comparison analysis to find differences between categories or groups.

1. Which product category has the highest number of purchases?
2. Which shopping mall has the highest number of transactions?
3. Which product category generates the highest total sales?
4. Who spends more in total: Male or Female customers?
5. Which payment method is used the most?
6. Which shopping mall generates the highest total sales?
7. Which product category has the highest average price?
8. Which shopping mall has the highest average customer spending?

---

# Part 3: Distribution Analysis

Use distribution analysis to understand how values are spread.

1. How are customer ages distributed?
2. What age range contains most customers?
3. How are product prices distributed?
4. Are most products low-priced, medium-priced, or expensive?
5. How is the purchased quantity distributed?
6. What quantity is purchased most often?
7. How is customer spending per transaction distributed?
8. Are most transactions small or large?
9. How are transactions distributed between Male and Female customers?

---

# Part 4: Relationship Analysis

Use relationship analysis to explore connections between two variables.

1. Is there a relationship between customer age and total spending?
2. Do older customers spend more money?
3. Is there a relationship between purchased quantity and total spending?
4. Does buying more items usually increase total spending?
5. Is there a relationship between product price and total spending?
6. Do expensive products always create higher transaction values?
7. Is there a relationship between customer age and purchased quantity?

---

# Part 5: Composition Analysis

Use composition analysis to understand how each part contributes to the whole.

1. What percentage of transactions comes from Male and Female customers?
2. What percentage of transactions uses each payment method?
3. What percentage of total sales comes from each product category?
4. What percentage of total sales comes from Male and Female customers?
5. What percentage of total sales comes from each shopping mall?
6. Which product category contributes the largest share of total sales?

---

# Final Challenge

Imagine that you are a **Data Analyst** working for a retail company.

Choose four important questions from the dataset:

* One **Comparison** question
* One **Distribution** question
* One **Relationship** question
* One **Composition** question

For each question:

1. Choose the appropriate chart.
2. Create the visualization.
3. Write one simple insight from the chart.
4. Explain why you selected this chart.
