# Columns understanding

### Education

```python
data['Education'] = data['Education'].replace('2n Cycle', 'Master')
```

```python
data['Education'].value_counts()
```

### Marital_status

```python
data['Marital_Status'].value_counts()
```

```python
print(data[data['Marital_Status'].isin(['YOLO', 'Absurd'])].to_string())
```

```python
data['Marital_Status'] = data['Marital_Status'].replace('Alone', 'Single')
```

```python
data = data[~data['Marital_Status'].isin(['YOLO', 'Absurd'])]
```

```python
data['Marital_Status'].value_counts()
```

### DT_Customer

```python
# data date converting
data['Dt_Customer'] = pd.to_datetime(data['Dt_Customer'], format='%d-%m-%Y')
```

```python
#separating date in three columns and dropping the original column
data['Year'] = data['Dt_Customer'].dt.year
data['Month'] = data['Dt_Customer'].dt.month
data['Day'] = data['Dt_Customer'].dt.day


data = data.drop(columns=['Dt_Customer'])

data.head()
```

### Year

```python
# year range
data['Year'].min(), data['Year'].max()
```

```python
# birth year to age
data['Age'] = data['Year'].max() - data['Year_Birth']

data = data.drop(columns=['Year_Birth'])
```

```python
data.head()
```

```python
print(data['Age'].describe())
print(data['Age'].sort_values(ascending=False).head(10))
```

```python
data[data['Age'] > 90]
```

```python
data = data[data['Age'] <= 90]
```

```python
data.info()
```

# Columns outlier

```python
sns.countplot(x='Education', data=data)
```

```python
sns.countplot(x='Marital_Status', data=data)
```

```python
numerical_cols = [col for col in data.columns if data[col].dtype in ['int64', 'float64']]
rows = len(numerical_cols) // 4 + 1

plt.figure(figsize=(16, rows * 3))
for i, column in enumerate(numerical_cols):
    plt.subplot(rows, 4, i + 1)
    plt.boxplot(data[column])
    plt.title(column)

plt.tight_layout()
plt.show()
```

```python
Q1 = data['Income'].quantile(0.25)
Q3 = data['Income'].quantile(0.75)
qr = Q3 - Q1
lower_bound = Q1 - 1.5 * qr
upper_bound = Q3 + 1.5 * qr

data['Income'] = data['Income'].apply(
    lambda x: upper_bound if x > upper_bound else (lower_bound if x < lower_bound else x)
)

plt.boxplot(data['Income'])
plt.title('Income column outliers (capped)')
plt.show()
```
