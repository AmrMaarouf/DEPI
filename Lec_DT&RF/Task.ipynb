```python
import numpy as np
import pandas as pd
import sklearn as sk
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from ucimlrepo import fetch_ucirepo

bank_marketing = fetch_ucirepo(id=222)

X = bank_marketing.data.features
y = bank_marketing.data.targets

bank_data = pd.concat([X, y], axis=1)

bank_data.head()
```

```python
bank_data.info()
```

```python
# converting yes no features to 1 and 0
bank_data['default'] = bank_data['default'].map({'yes': 1, 'no': 0})
```

```python
#drop the rows with job null values
bank_data = bank_data.dropna(subset=['job'])
```

```python
#null percentage in all columns
for col in bank_data.columns:
    null_percentage = (bank_data[col].isnull().sum() / len(bank_data)) * 100
    print(f"{col}: {null_percentage:.2f}%")
```
