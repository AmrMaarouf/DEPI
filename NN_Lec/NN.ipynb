```python
# 1) bulid model
model = keras.Sequential(
    [
        keras.Input(shape=(28,28)), # input data
        keras.layers.Flatten(), # 1) input layer with flatten
        keras.layers.Dense(16,activation='relu'), #hidden layer 1 , 16 neurons, 32 , 32 , 64, 128
        keras.layers.Dense(10, activation="softmax") # output layer
    ]

)
```

```python
model.summary()
# flatten (Flatten)  = 28*28 = 784
# hidden layer  = 784*16 + 16 = 12,560 (weights + biases )
# output layer =(16*10) + 10 = 170
```

```python
# 2) compile model
model.compile(
    optimizer ="adam", # weight-update methods (adam, sgd, rmsporp)
    loss = "sparse_categorical_crossentropy", # calc loss
    metrics = ["accuracy"] #
)
```

```python
# 3) train model
history = model.fit(
      x_train,
    y_train,
  epochs = 5,
  validation_data= (x_val,y_val)
)
```

```python
# plot history
plt.plot(history.history['loss'], label='train loss')
plt.plot(history.history['val_loss'], label='val loss')
plt.legend()
plt.show()
```

```python

test_loss , test_acc = model.evaluate(x_test,y_test)
print("test_acc: ", test_acc)
print("test_loss: ", test_loss)
```

```python
from tensorflow import keras

x_train_c = x_train.reshape(-1, 28, 28, 1)
x_val_c = x_val.reshape(-1, 28, 28, 1)
x_test_c = x_test.reshape(-1, 28, 28, 1)

model_cnn = keras.Sequential([
    keras.layers.Input((28, 28, 1)),
    keras.layers.RandomRotation(0.1),
    keras.layers.RandomTranslation(0.1, 0.1),
    keras.layers.Conv2D(32, 3, activation="relu"),
    keras.layers.MaxPooling2D(),
    keras.layers.Conv2D(64, 3, activation="relu"),
    keras.layers.MaxPooling2D(),
    keras.layers.Flatten(),
    keras.layers.Dropout(0.3),
    keras.layers.Dense(10, activation="softmax"),
])

model_cnn.compile(optimizer="adam",
                  loss="sparse_categorical_crossentropy",
                  metrics=["accuracy"])

model_cnn.fit(x_train_c, y_train, epochs=10,
              validation_data=(x_val_c, y_val))

print(model_cnn.evaluate(x_test_c, y_test, verbose=0))
```
